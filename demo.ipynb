# Fantastic Four: Sampling and Reliability

Run these cells in order using a CPU runtime. All data and results are simulated. The demo explains our proposed comparison; it does not use patient images or a pretrained pathology model.

```python
from pathlib import Path
import subprocess
import sys

repo = Path("/content/fantastic-four-demo")
if not repo.exists():
    subprocess.run(["git", "clone", "https://github.com/lukeblevins/fantastic-four-demo.git", str(repo)], check=True)
sys.path.insert(0, str(repo))
print("Demo files are ready.")
```

```python
from IPython.display import display
from google.colab import output
import time
import urllib.request

server = subprocess.Popen([sys.executable, "app.py"], cwd=repo)
for attempt in range(30):
    try:
        urllib.request.urlopen("http://127.0.0.1:8765", timeout=1).close()
        break
    except OSError:
        time.sleep(0.2)
else:
    raise RuntimeError("The demo server did not start.")
output.serve_kernel_port_as_iframe(8765, height=1100)
```
